##   <span style="color:blue">¿Qué es una Base de Datos en 2026?</span>

Una base de datos NO es simplemente una tabla con filas y columnas.

Formalmente:

Una base de datos es un sistema organizado para almacenar, gestionar, consultar y garantizar integridad de datos bajo restricciones formales y transaccionales.

En la nube moderna:
+ Persistencia
+ Consistencia
+ Concurrencia
+ Seguridad
+ Escalabilidad

##   <span style="color:blue">El Modelo Relacional</span>  

Propuesto por Edgar F. Codd (1970).

Una tabla es formalmente una relación:
+ Conjunto de tuplas
+ Sin orden
+ Sin duplicados
+ Con dominio definido

Conceptos clave:
+ Clave primaria
+ Clave foránea
+ Integridad referencial
+ Normalización

##   <span style="color:blue">Arquitectura Moderna</span>  

En entornos cloud actuales:

```text
Aplicación (Python / API)
        ↓
PostgreSQL (OLTP)
        ↓
Data Lake (S3 / RustFS)
        ↓
Motor Analítico (Dask / DuckDB)
```

La base de datos relacional es el núcleo transaccional.


##   <span style="color:blue">Implementación Práctica con PostgreSQL</span> 

 Usaremos SQLite para ejemplo local simple.

###   Crear Base de Datos en Python

````python
import sqlite3

# Crear base de datos en memoria
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE estudiante (
    id INTEGER PRIMARY KEY,
    nombre TEXT NOT NULL,
    edad INTEGER,
    programa TEXT
)
""")

conn.commit()
````

###  Insertar Datos

````python
cursor.execute("INSERT INTO estudiante (nombre, edad, programa) VALUES (?, ?, ?)",
               ("Laura", 22, "Ciencia de Datos"))

cursor.execute("INSERT INTO estudiante (nombre, edad, programa) VALUES (?, ?, ?)",
               ("Carlos", 24, "Ingeniería"))

conn.commit()
````

### Consultar Datos (SQL)

````python
cursor.execute("SELECT * FROM estudiante")
resultados = cursor.fetchall()

for fila in resultados:
    print(fila)
````

 ### Operaciones CRUD

CRUD significa:
+ Create
+ Read
+ Update
+ Delete

#### Update

````python
cursor.execute("UPDATE estudiante SET edad = 23 WHERE nombre = 'Laura'")
conn.commit()
````

#### Delete

````python
cursor.execute("DELETE FROM estudiante WHERE nombre = 'Carlos'")
conn.commit()
````

### Integridad y restricciones

````python
cursor.execute("""
CREATE TABLE curso (
    id INTEGER PRIMARY KEY,
    nombre TEXT NOT NULL UNIQUE
)
""")
````

Tipos de restricciones:
+ PRIMARY KEY
+ UNIQUE
+ NOT NULL
+ FOREIGN KEY
+ CHECK


### Relaciones entre tablas

````python
cursor.execute("""
CREATE TABLE matricula (
    estudiante_id INTEGER,
    curso_id INTEGER,
    FOREIGN KEY (estudiante_id) REFERENCES estudiante(id),
    FOREIGN KEY (curso_id) REFERENCES curso(id)
)
""")
````

Esto implementa integridad referencial.

### Bases de Datos vs Data Lakes

##   <span style="color:blue">Normalización Formal</span> 

$La normalización es un proceso formal para:

+ Eliminar redundancia
+ Reducir anomalías de actualización
+ Garantizar consistencia lógica

Se fundamenta en:
+ Dependencias funcionales
+ Teoría de claves
+ Teoría de conjuntos

⸻

### Dependencias Funcionales

Definición formal:

Sea una relación R.

Decimos que existe una dependencia funcional:

$$X \rightarrow Y$$

si para cualquier par de tuplas $t_1, t_2 \in R$:

$$t_1[X] = t_2[X] \Rightarrow t_1[Y] = t_2[Y]$$

Es decir: $X$ determina $Y$.

⸻

### Primera Forma Normal (1FN)

Una relación está en 1FN si:
+ Todos los atributos son atómicos.
+ No existen listas ni estructuras anidadas.

Ejemplo incorrecto:

| id | nombre | telefonos        |
|----|--------|------------------|
| 1  | Ana    | 300..., 310...   |

Debe separarse en otra tabla.

⸻

### Segunda Forma Normal (2FN)

Requiere:
+ Estar en 1FN.
+ No tener dependencias parciales respecto a claves compuestas.

Formalmente:

Si la clave es $(A,B)$, ningún atributo no-clave debe depender solo de $A$ o solo de $B$.

⸻

### Tercera Forma Normal (3FN)

Requiere:
+ Estar en 2FN.
+ No tener dependencias transitivas.

Si:

$$A \rightarrow B \quad \text{y} \quad B \rightarrow C$$

Entonces $A$ no debe determinar $C$ indirectamente en la misma tabla.

⸻

### Forma Normal de Boyce-Codd (BCNF)

Más estricta que 3FN.

Una relación está en BCNF si:

Para toda dependencia funcional:

$$X \rightarrow Y$$

$X$ es superclave.

⸻

##   <span style="color:blue">Comparativa Formal: Relacional vs NoSQL</span> 

### Modelo Relacional

Propiedades:
+ Esquema fijo
+ ACID
+ Integridad referencial
+ Teoría matemática formal

Ejemplos:
+ PostgreSQL
+ MySQL
+ SQL Server

⸻

### Modelo NoSQL

Tipos principales:
+ Documental (MongoDB)
+ Key-Value (Redis)

Características:
+ Esquema flexible

### ACID vs BASE

ACID:
+ Atomicidad
+ Consistencia
+ Aislamiento
+ Durabilidad

BASE:
+ Basically Available
+ Soft state
+ Eventual consistency

⸻

###  ¿Cuándo usar cada uno?

| Caso                          | Relacional | NoSQL |

| Alta escalabilidad global     | ⚠          | ✔      |
| Datos altamente estructurados | ✔          | ⚠      |
| Eventos masivos tipo logs     | ❌         | ✔      |

##   <span style="color:blue">Álgebra Relacional (Nivel Formal)</span> 

El modelo relacional no se basa en SQL.
Se basa en álgebra relacional.

SQL es una implementación declarativa.

⸻

### Operadores Fundamentales

Sea una relación R:
	1.	Selección (σ)
	2.	Proyección (π)
	3.	Unión (∪)
	4.	Diferencia (−)
	5.	Producto cartesiano (×)
	6.	Renombramiento (ρ)

⸻

### Selección (σ)

$$\sigma_{edad > 20}$$ 
(Estudiante)

Equivale a:

```sql
SELECT * FROM estudiante WHERE edad > 20;
```

⸻

### Proyección (π)

$$\pi_{nombre, edad}$$ 
(Estudiante)

Equivale a:
```sql
SELECT nombre, edad FROM estudiante;
```

⸻

### Join como Operación Derivada

Un JOIN se define como:

$$R \bowtie S = \sigma_{condicion}(R \times S)$$

Es decir:

1.	Producto cartesiano
2.	Selección por condición

⸻

### Cierre Algebraico

El álgebra relacional es cerrada:

El resultado de cualquier operación es otra relación.

Esto permite:
+ Composición
+ Optimización formal
+ Planificación de consultas

⸻
